```python
# ============================================================
# TIME SERIES ANALYSIS - JENA CLIMATE DATASET
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


# ------------------------------------------------------------
# 1. PLOT MULTIPLE VARIABLES
# ------------------------------------------------------------

# Temperature
plt.figure(figsize=(15, 5))
plt.plot(df.index, df["T (degC)"])
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Over Time")
plt.show()


# Atmospheric Pressure
plt.figure(figsize=(15, 5))
plt.plot(df.index, df["p (mbar)"])
plt.xlabel("Time")
plt.ylabel("Pressure (mbar)")
plt.title("Atmospheric Pressure Over Time")
plt.show()


# Relative Humidity
plt.figure(figsize=(15, 5))
plt.plot(df.index, df["rh (%)"])
plt.xlabel("Time")
plt.ylabel("Relative Humidity (%)")
plt.title("Relative Humidity Over Time")
plt.show()


# ------------------------------------------------------------
# 2. IDENTIFY LONG-TERM TREND
# ------------------------------------------------------------

# Calculate monthly average temperature
monthly_temp = df["T (degC)"].resample("ME").mean()

plt.figure(figsize=(15, 5))
plt.plot(monthly_temp)
plt.xlabel("Time")
plt.ylabel("Average Temperature (°C)")
plt.title("Monthly Average Temperature - Long-Term Trend")
plt.show()


# ------------------------------------------------------------
# 3. YEARLY TEMPERATURE PATTERN
# ------------------------------------------------------------

# Select one complete year
temp_2015 = df.loc["2015", "T (degC)"]

plt.figure(figsize=(15, 5))
plt.plot(temp_2015)
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Temperature During 2015")
plt.show()


# ------------------------------------------------------------
# 4. MONTHLY SEASONAL PATTERN
# ------------------------------------------------------------

# Calculate average temperature for each month
monthly_seasonality = df["T (degC)"].groupby(
    df.index.month
).mean()

month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(12, 5))

plt.plot(
    month_names,
    monthly_seasonality.values,
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Average Temperature (°C)")
plt.title("Monthly Temperature Seasonality")

plt.show()


# ------------------------------------------------------------
# 5. DAILY SEASONALITY
# ------------------------------------------------------------

# Calculate average temperature for each hour of the day
hourly_seasonality = df["T (degC)"].groupby(
    df.index.hour
).mean()

plt.figure(figsize=(12, 5))

plt.plot(
    hourly_seasonality.index,
    hourly_seasonality.values,
    marker="o"
)

plt.xlabel("Hour of Day")
plt.ylabel("Average Temperature (°C)")
plt.title("Daily Temperature Seasonality")

plt.xticks(range(0, 24))

plt.show()


# ------------------------------------------------------------
# 6. IDENTIFY SAMPLING FREQUENCY
# ------------------------------------------------------------

# Calculate difference between consecutive timestamps
time_difference = df.index.to_series().diff()

print("Most common time differences:")
print(time_difference.value_counts().head())


# ------------------------------------------------------------
# 7. CALCULATE THE MOST COMMON SAMPLING INTERVAL
# ------------------------------------------------------------

most_common_difference = time_difference.value_counts().idxmax()

print("\nMost common sampling interval:")
print(most_common_difference)


# Convert interval into minutes
sampling_minutes = most_common_difference.total_seconds() / 60

print("\nSampling interval in minutes:")
print(sampling_minutes)


# ------------------------------------------------------------
# 8. LET PANDAS INFER THE FREQUENCY
# ------------------------------------------------------------

# Use a smaller continuous section because the full dataset
# may contain missing observations.
frequency = pd.infer_freq(df.index[:1000])

print("\nInferred sampling frequency:")
print(frequency)


# ------------------------------------------------------------
# 9. VISUALIZE THE 10-MINUTE SAMPLING
# ------------------------------------------------------------

# Take a small section of the dataset
sample = df.loc["2015-01-01 00:00:00":"2015-01-01 06:00:00"]

plt.figure(figsize=(15, 5))

plt.plot(
    sample.index,
    sample["T (degC)"],
    marker="o"
)

plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Temperature with 10-Minute Sampling")

plt.show()


# ------------------------------------------------------------
# 10. CHECK THE FIRST FEW TIMESTAMPS
# ------------------------------------------------------------

print("\nFirst 10 timestamps:")
print(df.index[:10])


# ------------------------------------------------------------
# 11. CHECK WHETHER THE DATA IS MULTIVARIATE
# ------------------------------------------------------------

print("\nNumber of variables:", len(df.columns))

print("\nVariables:")
for column in df.columns:
    print("-", column)
```
